```python
import subprocess
from pathlib import Path

graphql_path = Path("apitypes.graphql")
if not graphql_path.exists():
    raise FileNotFoundError(f"{graphql_path} not found")

graphql_registry = graphql_path.read_text()

#call ollama
import ollama

def analyse_graphql_query(user_query: str, graph_registry: str, model: str = "llama2"):
    prompt = f"""
You are given a GraphQL type registry and a user query in natural language.
Return the best matching GraphQL type call from the registry that fits the user query.

GraphQL registry:
{graph_registry}

User query:
{user_query}

Instructions:
- Analyze the GraphQL registry.
- Retrieve the semantically matching type call in the graph that best suits the user query.
- Use only types, fields, and operations present in the registry.
- Prefer the closest existing GraphQL type or query.
- Respond with the matching GraphQL type or query structure.
"""
    #print(f"Prompt: {prompt}")
    return prompt
```

```python
# Call the local Ollama instance
def call_ollama(prompt: str):
    response = ollama.chat(
        model='llama3.2',
        messages=[{
            'role': 'user',
            'content': prompt
        }]
    )

    # Print the response text
    #print(response['message']['content'])
    return response['message']['content']
```

```python
user_query = "show me last 10 cash deposits in the bank"
prompt = analyse_graphql_query(user_query, graphql_registry)
response = call_ollama(prompt)
print(response)
```
